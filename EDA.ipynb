```python
trans_dis = df.groupby("Transport_Mode")["Disruption_Occurred"].sum()
```

```python
plt.figure(figsize=(10,6))
plt.bar(trans_dis.index, trans_dis.values)
plt.title("Disruptions by Transport Mode")
plt.xlabel("Transport Mode")
plt.ylabel("Number of Disruptions")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df.drop(columns="Month", inplace=True)
```

```python
df.head()
```

```python
df = df.sort_values("Date").reset_index(drop=True)
```

```python
df["Month"] = pd.to_datetime(df["Date"]).dt.month
df["Day"] = df["Date"].dt.day

pivot = df.pivot_table(values="Disruption_Occurred", index="Day", columns="Month", aggfunc="sum", fill_value=0)

plt.figure(figsize=(12,8))
sns.heatmap(pivot, cmap="YlGnBu", fmt="d")
plt.title("Disruptions by Day and Month")
plt.xlabel("Month")
plt.ylabel("Day")
plt.show()
```

```python
df["month_name"] = pd.to_datetime(df["Date"]).dt.strftime("%B")
df.head()
```

```python
df.groupby("Month")["Disruption_Occurred"].sum().plot(kind='bar')
```

```python
df.drop(columns=["Rolling_Disruptions"], inplace=True)
```

```python
df["Lead_Time_Days"].describe()
```

```python
df.groupby("Product_Category")["Disruption_Occurred"].sum().plot(kind='bar')
```

```python
df.groupby("Product_Category")["Lead_Time_Days"].mean()
```

```python

total_disruption = (df["Disruption_Occurred"].sum() / len(df)) * 100

total_disruption
```

```python
disruptive_destination = df.groupby("Destination_Port")["Disruption_Occurred"].sum().sort_values(ascending=False)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=disruptive_destination.values, y=disruptive_destination.index, palette="viridis")
for i, v in enumerate(disruptive_destination.values):
    plt.text(v + 2, i, str(v))
plt.title("Disruptions by Destination Port")
plt.xlabel("Number of Disruptions")
plt.ylabel("Destination Port")
plt.tight_layout()
plt.show()
```
